```python
#EXPERIMENT 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data =pd.read_excel("sku_distributions 2.xlsx")
data.head()
```

```python
resistor_data =np.array(data['resistors'])
resistor_data
```

```python
data.shape
```

```python
data.info()
```

```python
from scipy.stats import norm,normaltest,kstest
```

```python
mean =resistor_data.mean()
sd =resistor_data.std()
kstest(resistor_data, 'norm', args=(mean,sd))
```

```python
import seaborn as sns
```

```python
capacitor_data =np.array(data['capacitors'])
```

```python
sns.kdeplot(resistor_data,label='distribution of resistor',fill=True, color='r',alpha=0.1)
plt.legend()
```

```python
import statsmodels.api as sm
```

```python
sm.qqplot(resistor_data, norm, fit=True, line='45')
plt.title('Q-Q plot for resistor data')
```
