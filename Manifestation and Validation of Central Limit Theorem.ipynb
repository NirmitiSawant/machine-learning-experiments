```python
#Validity of  central limit theorem
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

sample_size=100
total_sample=30000
#generate sample data for tem column
population_data=dataset['Temperature']
sample =np.random.choice(population_data ,(total_sample ,sample_size))
```

```python
sample
```

```python
sample[0]
```

```python
#calculating mean
sample_mean=np.mean(sample ,axis=1)
sample_mean.shape
```

```python
#plotting original temp data using histogram
plt.figure(figsize=(18,6))
plt.hist(population_data ,bins=100,color='red')
plt.show()
```

```python
#plotting sample distribution of means
plt.figure(figsize=(18,6))
plt.hist(sample_mean ,bins=100,color='blue')
plt.show()
```
