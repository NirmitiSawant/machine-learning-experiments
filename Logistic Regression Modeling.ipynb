```python
data['Q1'] = data['Q1'].replace({'o': 0, '0':0})
data['Q2'] = data['Q2'].replace({'o': 0, '0':0})
data['Q3'] = data['Q3'].replace({'o': 0, '0':0})
data['Q4'] = data['Q4'].replace({'o': 0, '0':0})
data['Q5'] = data['Q5'].replace({'o': 0, '0':0})
data['Q6'] = data['Q6'].replace({'o': 0, '0':0})
data['Q7'] = data['Q7'].replace({'o': 0, '0':0})
data['Q8'] = data['Q8'].replace({'o': 0, '0':0})
data['Q9'] = data['Q9'].replace({'o': 0, '0':0})
data['Q10'] = data['Q10'].replace({'o': 0, '0':0})
data['Q11'] = data['Q11'].replace({'o': 0, '0':0})
data['Q12'] = data['Q12'].replace({'o': 0, '0':0})
data['Q13'] = data['Q13'].replace({'o': 0, '0':0})
data['Q14'] = data['Q14'].replace({'o': 0, '0':0})
data['Q15'] = data['Q15'].replace({'o': 0, '0':0})
data['Q16'] = data['Q16'].replace({'o': 0, '0':0})
data['Q17'] = data['Q17'].replace({'o': 0, '0':0})
data['Q18'] = data['Q18'].replace({'o': 0, '0':0})
data['Q19'] = data['Q19'].replace({'o': 0, '0':0})
data['Q20'] = data['Q20'].replace({'o': 0, '0':0})
```

```python
x = data.drop(columns='Result') # take all question columns (20 questions score) into X varia
y = data['Result']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=44,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_predicted_train = model.predict(x_train) # provide input t
y_predicted_train
```

```python
y_test
```

```python
y_predicted_test = model.predict(x_test) 
y_predicted_test
```

```python
pred_prob = model.predict(x_test)
pred_prob

```

```python
(pred_prob > 0.5).astype(int)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
print(classification_report(y_train,y_predicted_train))
```

```python
print(classification_report(y_test,y_predicted_test))
```

```python
confusion_matrix(y_test, y_predicted_test)
```

```python
cm = confusion_matrix(y_test, y_predicted_test)
sns.heatmap(cm, annot=True, linewidths=0.3, linecolor='white',
cmap='coolwarm',fmt='d')
plt.xlabel('Prediction Labels')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn import metrics
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
auc = roc_auc_score(y_test, pred_prob)
print("Area under curve {:.2f}".format(auc))
```

```python
fpr, tpr, thresh = roc_curve(y_test, pred_prob, pos_label=1)
plt.plot(fpr, tpr, linestyle = '-.',color='blue',label= 'auc = '+str(auc))
plt.plot([0, 1], [0, 1], linestyle='--',color='Red')
plt.title('ROC Curves')
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt. legend(loc='lower right')
plt.show()
```
